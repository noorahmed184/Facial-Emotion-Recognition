```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split

from keras.callbacks import EarlyStopping
from keras.models import Sequential
from keras.layers import Dense, Input, Activation, Flatten
from keras.utils import to_categorical

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
train_folder = train_data
test_folder = os.path.join(path, "test")
```

```python
emotions = os.listdir(train_folder)
```

```python
from tensorflow.keras.utils import load_img
import matplotlib.pyplot as plt

for i, emotion in enumerate(emotions):
    emotion_folder = os.path.join(train_folder, emotion)
    image_names = os.listdir(emotion_folder)

    image_path = os.path.join(emotion_folder, image_names[0])
    img = load_img(image_path, color_mode="grayscale")

    plt.subplot(2, 4, i + 1)
    plt.imshow(img, cmap="gray")
    plt.title(emotion)

plt.show()
```

```python
import tensorflow as tf
# size of the photos and number of batch we ttain each time 
img_size = (48, 48)
batch_size = 32
```

```python
train_data = tf.keras.utils.image_dataset_from_directory(
    train_folder,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=img_size,
    color_mode="grayscale",
    batch_size=batch_size
)

val_data = tf.keras.utils.image_dataset_from_directory(
    train_folder,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=img_size,
    color_mode="grayscale",
    batch_size=batch_size
)
```

```python
test_data = tf.keras.utils.image_dataset_from_directory(
    test_data,
    image_size=img_size,
    color_mode="grayscale",
    batch_size=batch_size,
)
```

```python
# do the normalization to be the values in same range 0 or 1
train_data = train_data.map(lambda x, y: (x / 255, y))
val_data = val_data.map(lambda x, y: (x / 255, y))
test_data = test_data.map(lambda x, y: (x / 255, y))
```

```python
from keras.utils import set_random_seed
import random
from keras.layers import Dense, Activation, Flatten
from keras.models import Sequential

seed = 9999
set_random_seed(seed)

model = Sequential([
                    Flatten(input_shape=(48, 48, 1)), 
                    Dense(120, activation='relu'), 
                    Dense(7, activation='softmax') 
                    ])


model.summary()
```

```python
model.weights
```

```python
model.compile(loss="sparse_categorical_crossentropy", metrics=["accuracy"])

history = model.fit(train_data, validation_data=val_data, epochs=10)
```

```python
history_df = pd.DataFrame(history.history)
history_df.head()
```

```python
history_df[['accuracy', 'val_accuracy']].plot();
```

```python
# see the accuracy of model and how many loss 
test_loss, test_accuracy = model.evaluate(test_data)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import numpy as np
import matplotlib.pyplot as plt

y_true = np.concatenate([
    labels.numpy()
    for images, labels in test_data
])

y_pred_test = np.argmax(
    model.predict(test_data, verbose=0),
    axis=1
)

class_names = [
    "angry", "disgust", "fear", "happy",
    "neutral", "sad", "surprise"
]

ConfusionMatrixDisplay.from_predictions(
    y_true,
    y_pred_test,
    labels=range(7),
    display_labels=class_names,
    xticks_rotation=45
)

plt.show()
```

```python
import numpy as np
from sklearn.metrics import classification_report

y_true = np.concatenate([
    labels.numpy()
    for images, labels in test_data
])

y_pred = np.argmax(model.predict(test_data), axis=1)
```

```python
class_names = [
    "angry",
    "disgust",
    "fear",
    "happy",
    "neutral",
    "sad",
    "surprise"
]
print(classification_report(
    y_true,
    y_pred,
    target_names=class_names
))
```
