```python
import tensorflow as tf
# size of the photos and number of batch we ttain each time
img_size = (48, 48)
batch_size = 32
```

```python
# praper data for train and validation
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    train_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=img_size,
    color_mode="grayscale",
    batch_size=batch_size
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    train_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=img_size,
    color_mode="grayscale",
    batch_size=batch_size
)
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    test_path,
    seed=42,
    image_size=img_size,
    color_mode="grayscale",
    batch_size=batch_size,
)
```

```python
# for normalization divade by 225 to becom from 0 and 1 ,and for training make faster

train = test_ds.map(lambda x, y: (x / 255, y))
val = val_ds.map(lambda x, y: (x / 255, y))
test = test_ds.map(lambda x, y: (x / 255, y))
```

```python
from keras.utils import set_random_seed
import random
from keras.layers import Dense, Activation, Flatten
from keras.models import Sequential

seed = 9999
set_random_seed(seed)

model = Sequential([
                    Flatten(input_shape=(48, 48, 1)), # make pix flat becuse the dense exapt only flat
                    Dense(120, activation='relu'), # make nagative number 0
                    Dense(7, activation='softmax') # make output sum = 1 and must highist one is the one
                    ])


model.summary()
```
